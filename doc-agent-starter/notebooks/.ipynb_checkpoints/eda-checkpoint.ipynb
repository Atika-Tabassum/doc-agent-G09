# A1 — Corpus Exploration
Explore the scanned corpus in `data/raw/`: page/word counts, scan quality, script and font variety.

Corpus: *Rabindra Rachanabali*, Volume 25 (Bengali), sourced from Internet Archive
(`RABINDRARACHANABALI` item, `data/provenance.md` has full source/license details).


```python
import sys
from pathlib import Path

# Make src/ importable (repo convention: pyproject pythonpath = ["src"])
ROOT = Path.cwd()
if (ROOT / "src").exists():
    sys.path.insert(0, str(ROOT / "src"))
elif (ROOT.parent / "src").exists():
    ROOT = ROOT.parent
    sys.path.insert(0, str(ROOT / "src"))

RAW_DIR = ROOT / "data" / "raw"
print("Project root:", ROOT)
print("Raw data dir:", RAW_DIR, "exists:", RAW_DIR.exists())
```

## 1. Page count

Walk `data/raw/` directly (this notebook explores the *raw* corpus before any
pipeline stage runs, so it does not depend on `ingest/loader.py` being implemented).
Each subfolder under `data/raw/` is one `doc_id`; each subfolder currently holds
the raw page images produced by `scripts/get_data.sh`.


```python
import pandas as pd

IMG_EXTS = {".png", ".jpg", ".jpeg", ".tif", ".tiff"}

records = []
for doc_dir in sorted(p for p in RAW_DIR.iterdir() if p.is_dir()):
    pages = sorted(p for p in doc_dir.iterdir() if p.suffix.lower() in IMG_EXTS)
    for p in pages:
        records.append({"doc_id": doc_dir.name, "path": str(p), "filename": p.name})

pages_df = pd.DataFrame(records)
n_docs = pages_df["doc_id"].nunique() if not pages_df.empty else 0
n_pages = len(pages_df)

print(f"Documents (doc_id folders): {n_docs}")
print(f"Total pages: {n_pages}")
pages_df.groupby("doc_id").size().rename("pages").to_frame()
```

**Target check:** the corpus must have **≥300 pages**. See the cell output above —
if `n_pages` is below 300, more volumes/documents need to be added via `scripts/get_data.sh`
before this satisfies the A1/A2 corpus size requirement in `configs/task.yaml`.


## 2. Word count (post-OCR estimate)

The real OCR stage (`vision/ocr.py`) is not implemented yet at this point in the
pipeline, so an exact word count isn't available. We estimate it two ways:

1. **If `pytesseract` + the Tesseract binary are installed**, we OCR a random
   sample of pages and extrapolate to the full corpus.
2. **Otherwise**, we fall back to a documented per-page estimate (e.g. from the
   source's own metadata/known page density for this script), clearly labelled
   as an estimate, not a measurement.

Replace the fallback constant once real OCR (Person B's stage) is available, and
re-run this cell for the true count in `notebooks/kb_demo.ipynb`.


```python
import random

SAMPLE_SIZE = min(15, n_pages)
word_count_estimate = None
method = None

try:
    import pytesseract
    from PIL import Image

    sample_paths = random.sample(list(pages_df["path"]), SAMPLE_SIZE) if n_pages else []
    sample_word_counts = []
    for p in sample_paths:
        img = Image.open(p)
        text = pytesseract.image_to_string(img, lang="ben")  # Bengali; adjust lang per corpus
        sample_word_counts.append(len(text.split()))

    if sample_word_counts:
        avg_words_per_page = sum(sample_word_counts) / len(sample_word_counts)
        word_count_estimate = int(avg_words_per_page * n_pages)
        method = f"OCR sample (n={len(sample_word_counts)} pages, tesseract)"
except ImportError:
    pass

if word_count_estimate is None:
    # Fallback: documented rough estimate, NOT a measurement.
    # TODO(Person A): replace with a real figure once vision/ocr.py (Person B) is implemented,
    # or install `pytesseract` + tesseract-ocr + the `ben` language pack to enable the OCR-sample path above.
    ASSUMED_WORDS_PER_PAGE = 250  # typical for a printed Bengali literary page at this trim size
    word_count_estimate = ASSUMED_WORDS_PER_PAGE * n_pages
    method = f"fallback heuristic ({ASSUMED_WORDS_PER_PAGE} words/page assumed, not measured)"

print(f"Estimated total words: {word_count_estimate:,}")
print(f"Method: {method}")
```

you report there.


## 3. Scan quality samples

Show a few example pages and basic quality signals: brightness, contrast, and a
simple blur/noise proxy (edge-detail via Laplacian variance — lower usually means
blurrier/noisier scans). No OpenCV dependency needed; using PIL + numpy only.


```python
import numpy as np
from PIL import Image, ImageFilter
import matplotlib.pyplot as plt

def quality_signals(path):
    img = Image.open(path).convert("L")  # grayscale
    arr = np.asarray(img, dtype=np.float32)
    brightness = arr.mean()
    contrast = arr.std()
    edges = np.asarray(img.filter(ImageFilter.FIND_EDGES), dtype=np.float32)
    sharpness_proxy = edges.var()  # crude stand-in for Laplacian variance
    return brightness, contrast, sharpness_proxy

sample_n = min(6, n_pages)
sample_rows = pages_df.sample(sample_n, random_state=42) if n_pages else pages_df

fig, axes = plt.subplots(2, 3, figsize=(14, 8))
quality_rows = []
for ax, (_, row) in zip(axes.flat, sample_rows.iterrows()):
    img = Image.open(row["path"])
    ax.imshow(img, cmap="gray")
    ax.set_title(row["filename"], fontsize=9)
    ax.axis("off")
    b, c, s = quality_signals(row["path"])
    quality_rows.append({"filename": row["filename"], "doc_id": row["doc_id"],
                          "brightness": round(b, 1), "contrast": round(c, 1),
                          "sharpness_proxy": round(s, 1)})

plt.tight_layout()
plt.show()

pd.DataFrame(quality_rows)
```

## 4. Script / font variety

The corpus is Bengali (Bangla script). Since this is a single scanned book series
(not a multi-source corpus), font variety is expected to be low — but check for
inconsistencies across volumes/sections (e.g. cover pages, publisher's front matter,
or footnotes may use a different type size/weight than body text).


```python
# Group pages by doc_id and show one representative page per doc_id
# to visually compare typesetting across documents/sub-volumes.
reps = pages_df.groupby("doc_id").first().reset_index()

fig, axes = plt.subplots(1, len(reps), figsize=(5 * len(reps), 6))
if len(reps) == 1:
    axes = [axes]
for ax, (_, row) in zip(axes, reps.iterrows()):
    img = Image.open(row["path"])
    ax.imshow(img, cmap="gray")
    ax.set_title(row["doc_id"], fontsize=10)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
